# 03 - Campaign Performance Analysis

This notebook performs comprehensive analysis of campaign performance.

## Objectives:
- Calculate key performance metrics
- Identify top-performing campaigns
- Analyze performance by dimensions (age, region, device, etc.)
- Time series analysis
- Generate insights and recommendations


```python
# Import libraries
import pandas as pd
import numpy as np
import sys

sys.path.append('../src')
from data_processing import load_data
from analysis import get_campaign_summary, get_top_performing_ads, get_campaign_performance_by_dimension
from utils import format_currency, format_percentage
```

```python
# Load cleaned data
data_path = '../data/processed/cleaned_campaign_data.csv'
df = load_data(data_path, file_type='csv')

print(f"Data shape: {df.shape}")
```

```python
# Get campaign summary
summary = get_campaign_summary(df)

print("=== CAMPAIGN SUMMARY ===")
for key, value in summary.items():
    if 'Spend' in key or 'Revenue' in key:
        print(f"{key}: {format_currency(value)}")
    elif 'Rate' in key or 'CTR' in key or 'ROI' in key:
        print(f"{key}: {format_percentage(value)}")
    else:
        print(f"{key}: {value:,.0f}")
```

# 02 - Data Cleaning

This notebook cleans and preprocesses the campaign data.

## Objectives:
- Remove duplicates
- Handle missing values
- Convert data types
- Standardize column names
- Calculate performance metrics
- Save cleaned data


```python
# Import libraries
import pandas as pd
import numpy as np
import sys

sys.path.append('../src')
from data_processing import load_data, clean_data, calculate_metrics
```

```python
# Load raw data
data_path = '../data/raw/campaign_data.csv'  # Update with your file path
df_raw = load_data(data_path, file_type='csv')

print(f"Raw data shape: {df_raw.shape}")
```

```python
# Clean the data
df_clean = clean_data(df_raw)

print(f"Cleaned data shape: {df_clean.shape}")
print(f"Rows removed: {len(df_raw) - len(df_clean)}")
```

```python
# Calculate metrics
df_metrics = calculate_metrics(df_clean)

print("Calculated metrics:")
metric_cols = [col for col in df_metrics.columns if col in ['CTR', 'CPC', 'CPM', 'Engagement_Rate', 'Conversion_Rate', 'ROI', 'ROAS']]
print(metric_cols)
```

```python
# Save cleaned data
output_path = '../data/processed/cleaned_campaign_data.csv'
df_metrics.to_csv(output_path, index=False)
print(f"Cleaned data saved to: {output_path}")
```
